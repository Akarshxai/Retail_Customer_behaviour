```python
import pandas as pd
```

```python
df=pd.read_csv("C:\\Users\\akars\\Downloads\\customer_shopping_behavior.csv")
df.head(10)
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
## Filling the missing values in the Review Rating column with median values of each category
df['Review Rating']=df.groupby('Category')['Review Rating'].transform(lambda x: x.fillna(x.median()))
```

```python
df.isnull().sum()
```

```python
##Cleaning the column names
df.columns=df.columns.str.lower()
df.columns=df.columns.str.replace(' ', '_')
df=df.rename(columns={'purchase_amount_(usd)':'purchase_amount'})
```

```python
df.columns
```

```python
## Create a column age_group based on the age column
df['age_group'] = pd.cut(df['age'], bins=[0, 18, 30, 45, 60, 100], labels=['child', 'teenager', 'adult', 'middle-aged', 'senior'])
df[['age', 'age_group']].head(10)
```

```python
df['frequency_of_purchases'].value_counts()
```

```python
## Creating a new column 'purchase_frequency_days' based on the 'frequency_of_purchases' column
frequency_mapping = {
    'Fortnightly': 14,
    'Weekly': 7,
    'Bi-Weekly': 14,
    'Monthly': 30,
    'Quarterly': 90,
    'Annually': 365,
    'Every 3 Months': 90
}
df['purchase_frequency_days'] = df['frequency_of_purchases'].map(frequency_mapping)
df[['frequency_of_purchases', 'purchase_frequency_days']].head(10)
```

```python
(df['discount_applied']==df['promo_code_used']).all()
```

```python
df=df.drop(['promo_code_used'], axis=1)
```

```python
df.to_csv("cleaned_data.csv", index=False)
```
